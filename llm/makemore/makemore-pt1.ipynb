## Why do we need one_hot instead of just numbers to find gradient
Because a neural network learns by computing **gradients** (how much to change each weight).

Gradients only make sense for **floating-point vectors**, not raw integers.

**Raw integers:**
```python
x = torch.tensor([5])  # just a number
# Can't backprop through this — what does "change 5 slightly" even mean?
# It's categorical, not a learned representation
```

**One-hot vectors:**
```python
x = torch.tensor([0, 0, 0, 1, 0, 0, 0])  # one-hot for index 3
# Now it's a float vector — gradient descent can nudge each component
# The network learns: "when I see this pattern, adjust my weights this way"
```

**Why different dimensions matter:**

The network's weight matrix `W` is `(27, 27)`. It expects input of shape `(batch, 27)`.

```python
# This fails:
logits = indices @ W  # (batch,) @ (27, 27) → shape mismatch

# This works:
logits = one_hot @ W  # (batch, 27) @ (27, 27) → (batch, 27) ✓
```

### Additionally

@ is matrix multiplication.

Shape rule:

**(a, b) @ (b, c) → (a, c)**

Inner dimensions must match and cancel. Outer dimensions stay.

````python
# Building full training set as two parallel tensors xs i.e 
# input character indices and ys target next character indices

# We don't use P or N,
# Instead, you create a training dataset: pairs of (input character, target next character)
# You train a neural network W to predict the target given the input
# W is one hot of 27 and target is whats predicted


# Step 1: Extract all bigrams into xs (input) and ys (target)
xs = []
ys = []

for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        xs.append(ix1)
        ys.append(ix2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)

print(f'xs shape: {xs.shape}')
print(f'ys shape: {ys.shape}')
print(f'First 10 xs: {xs[:10]}')
print(f'First 10 ys: {ys[:10]}')

# Step 2: Convert xs to one-hot
import torch.nn.functional as F

xenc = F.one_hot(xs, num_classes=27).float()
print(f'xenc shape: {xenc.shape}')
print(f'xenc dtype: {xenc.dtype}')
print(f'First one-hot vector: {xenc[0]}')

# Step 3: Verify each row of xenc has exactly one 1
print(f'Sum of each row (should be 1.0): {xenc.sum(1)[:10]}')

print(xenc.shape)
print(xenc[0])
````

````python
# Lets create a Seeded Generator that will predict next word
# This is the actual model we are gonna train


# W is the weight matrix — the thing the network learns.
# Each row of W is a "learned pattern" for what character typically follows a given input character.

g = torch.Generator().manual_seed(2147483647)
W = torch.randn(27,27)
W
````

````python
# Logits are raw, unnormalized network output scores — before softmax converts them to probabilities.
logits = xenc @ W
logits

# each item of logits are the random likehood of how much the next word index can be in terms of index of 27
# its like how much can A be in numbers

#NOTE
# logits[i] = network's prediction for what comes after xs[i]
````

````python
# We have logits from: logits = xenc @ W
# Shape: (batch_size, 27)

# Step 1: Exponentiate
counts = logits.exp()
# exp() makes all values positive
# larger logits → much larger exp values
# negative logits → small exp values (close to 0)

# Step 2: Normalize by row sum
probs = counts / counts.sum(1, keepdim=True)
# Divide each row by its sum → each row sums to 1
# Now each row is a valid probability distribution



# NOTE: This is Softmax what we use in ML or AI
# Manual:
probs = logits.exp() / logits.exp().sum(1, keepdim=True)

# Or built-in:
probs = F.softmax(logits, dim=1)
````

````python
# Phase 10: Evaluate the network's predictions

# Step 1: For a few examples, manually inspect
for i in range(5):
    # Get the input and target
    input_char = itos[xs[i].item()]
    target_char = itos[ys[i].item()]
    
    # Get the probability the network assigned to the target
    prob = probs[i, ys[i]].item()
    
    print(f'{input_char} → {target_char}: prob = {prob:.4f}')

# Step 2: Compute loss for all examples at once
# Extract probability of correct answer for each example
correct_probs = probs[range(len(ys)), ys]

print(f'correct_probs shape: {correct_probs.shape}')
print(f'correct_probs sample: {correct_probs[:10]}')
print(f'Min prob: {correct_probs.min():.6f}')
print(f'Max prob: {correct_probs.max():.6f}')

# Step 3: Compute negative log likelihood
log_probs = torch.log(correct_probs)
loss = -log_probs.mean()

print(f'loss: {loss.item():.4f}')

# High loss = network assigned low probability to correct answers = bad
````

````python
# Lets recreate W, but not turn on learning like gradient
# Phase 11: Actual gradient descent

# ========== CELL 1: Phase 11 Training ==========

# Create W with requires_grad=True
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27, 27), generator=g, requires_grad=True)

print(f'W shape: {W.shape}')
print(f'W requires_grad: {W.requires_grad}')

# Training loop
learning_rate = 50
losses = []

for iteration in range(1000):
    # ========== FORWARD PASS ==========
    logits = xenc @ W  # (228146, 27)
    
    # Softmax() applied
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    
    correct_probs = probs[range(len(ys)), ys]
    log_probs = torch.log(correct_probs)
    loss = -log_probs.mean()
    
    losses.append(loss.item())
    
    # ========== BACKWARD PASS ==========
    if W.grad is not None:
        W.grad.zero_()
    
    loss.backward()
    
    # ========== UPDATE WEIGHTS ==========
    W.data += -learning_rate * W.grad
    
    if iteration % 100 == 0:
        print(f'iteration {iteration}: loss = {loss.item():.4f}')

print(f'\nFinal loss: {loss.item():.4f}')
print(f'Expected (Phase 6/7): ~2.45')
````

````python
# ========== CELL 2: Plot training loss ==========

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(losses)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Training Loss Over Time')
plt.grid()
plt.show()
````

````python
# ========== CELL 3: Phase 12 - Sample from trained network ==========

g = torch.Generator().manual_seed(2147483647)

print('Generated names from trained network:\n')

for i in range(10):
    out = []
    ix = 0
    
    while True:
        xenc_single = F.one_hot(torch.tensor(ix), num_classes=27).float()
        logits = xenc_single @ W
        counts = logits.exp()
        probs = counts / counts.sum()
        ix = torch.multinomial(probs, num_samples=1, replacement=True, generator=g).item()
        out.append(itos[ix])
        
        if ix == 0:
            break
    
    print(''.join(out))
````

````python
# ========== CELL 4: Optional - L2 Regularization ==========

# L2 regularization adds a penalty for having large weights.
# total_loss = original_loss + λ * (W²).mean()
# Large weights can overfit — memorize the training data instead of learning general patterns.
#


# Create new W for L2 experiment
W_l2 = torch.randn((27, 27), generator=g, requires_grad=True)


learning_rate = 50
reg_strength = 0.01  # L2 regularization strength
losses_l2 = []

print('Training WITH L2 regularization:\n')

for iteration in range(1000):
    # Forward pass
    logits = xenc @ W_l2
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    correct_probs = probs[range(len(ys)), ys]
    log_probs = torch.log(correct_probs)
    loss = -log_probs.mean()
    
    # Add L2 regularization: penalize large weights
    reg_loss = reg_strength * (W_l2 ** 2).mean()
    total_loss = loss + reg_loss
    
    losses_l2.append(total_loss.item())
    
    # Backward pass
    if W_l2.grad is not None:
        W_l2.grad.zero_()
    
    total_loss.backward()
    
    # Update weights
    W_l2.data += -learning_rate * W_l2.grad
    
    if iteration % 100 == 0:
        print(f'iteration {iteration}: loss = {loss.item():.4f}, reg_loss = {reg_loss.item():.4f}, total = {total_loss.item():.4f}')

print(f'\nFinal total loss (with L2): {total_loss.item():.4f}')
print(f'Without L2: {losses[-1]:.4f}')
print(f'\nL2 keeps weights smaller (less overfit)')
````

````python
# ========== CELL 5: Compare L2 vs no L2 ==========

plt.figure(figsize=(12, 6))
plt.plot(losses, label='No L2', alpha=0.7)
plt.plot(losses_l2, label='L2 (0.01)', alpha=0.7)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Training Loss: With vs Without L2 Regularization')
plt.legend()
plt.grid()
plt.show()

print(f'Weight magnitude without L2: {(W ** 2).mean().sqrt().item():.4f}')
print(f'Weight magnitude with L2: {(W_l2 ** 2).mean().sqrt().item():.4f}')
print(f'\nL2 regularization penalizes large weights, keeping them smaller')
````
